## Imports and Given paramaters

```python
from scipy.integrate import solve_ivp
import numpy as np
import matplotlib.pyplot as plt


T0=330 # K temp of feed

Ftot=163 # Kmol h-1
FA0=0.9*Ftot
FB0=0
FI0=0.1*Ftot

v0=15.77
CA0=9.3 # mol L-1 = kmol m-3
CB0=0
CI0=FI0/v0

z0=[FA0,FB0,FI0,T0]
```

## rate related

```python
CPA=141 # J mol-1 K-1
CPB=141 # J mol-1 K-1
CPI=161 # J mol-1 K-1

Hr=-6900  # J mol-1 Enthalpy of reaction
Ea=65.7e-3 # J mol-1
R=8.314

k1=31.1 #Rate of reaction
T_r1=360 #temp of rate of reaction

kc2=3.03 # Equilibrium const
T_kc2=60+273.15 #At 60 C
```

```python
def isom_but(V,z,v_flow):

    '''
    Model of the adiabatic plug-flow reactor.

    State variables:
        z[0] = molar flow rate of A, kmol/h
        z[1] = molar flow rate of B, kmol/h
        z[2] = molar flow rate of inert I, kmol/h
        z[3] = temperature, K

    Returns:
    dFA/dV, dFB/dV, dFI/dV, dT/dV
    '''

    FA=z[0]
    FB=z[1]
    FI=z[2]
    T=z[3]

    CA=FA/v_flow
    CB=FB/v_flow

    k=k1*np.exp((-Ea/R)*((1/T)-(1/T_r1)))
    Kc=kc2*np.exp((-Hr/R)*((1/T)-(1/T_kc2)))
    r=-k*(CA-(CB/kc))
    rA=r
    dFAdV=rA
    dFBdV=-rA
    dFIdV=0

    # dTdV=((-Hr*-rA)-U*a*(T-Ta))/(FA*CPA+FB*CPB+FI*CPI) # If there was energy transfer/ transfer fluid
    dTdV=(-Hr*-rA)/(FA*CPA+FB*CPB+FI*CPI)

    return [dFAdV,dFBdV,dFIdV,dTdV]
```

## Solving differentials

```python
V_max=19 # vol
V_span=(0,V_max)
Steps=1000
V_array=np.linspace(0,V_max,Steps)
solution = solve_ivp(isom_but,t_span=V_span,y0=z0,t_eval=V_array,args=(v0,))
print(solution.success)
print(solution.message)
print(solution.t)
print(solution.y)
```

## Tabulated values

```python
V = solution.t
FA = solution.y[0]
FB = solution.y[1]
FI = solution.y[2]
T = solution.y[3]

k = k1 * np.exp((-Ea / R) * ((1 / T) - (1 / T_r1)))
Kc = Kc2 * np.exp((-Hr / R) * ((1 / T) - (1 / T_kc2)))
Xe = Kc / (1 + Kc)
X = 1 - FA / FA0

print(
    '    {0:12} {1:12} {2:14} {3:12} {4:12} {5:12} {6:12} {7:12} {8:12}'
    ' {9:12}'.format(
        'Volume', 'Conversion', 'F_A', 'F_B', 'F_I', 'T', 'k', 'Kc', 'Xe', 'X'
    )
)

for i in range(Steps):
  print(
      '{0:12.6f} {1:12.6f} {2:12.6f} {3:12.6f} {4:12.6f} {5:12.6f} {6:12.6f}'
      ' {7:12.6f} {8:12.6f} {9:12.6f}'.format(
          V[i], FA[i], FB[i], FI[i], T[i], k[i], Kc[i], Xe[i], X[i], X[i]
      )
  )
```
